```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
```

```python
iris = datasets.load_iris()
data = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df=data.iloc[:,1:3]
```

```python
mean1 = [np.random.uniform(df['sepal width (cm)'].min(), df['sepal width (cm)'].max()),
         np.random.uniform(df['petal length (cm)'].min(), df['petal length (cm)'].max())]

mean2 = [np.random.uniform(df['sepal width (cm)'].min(), df['sepal width (cm)'].max()),
         np.random.uniform(df['petal length (cm)'].min(), df['petal length (cm)'].max())]

mean3=[np.random.uniform(df['sepal width (cm)'].min(), df['sepal width (cm)'].max()),
         np.random.uniform(df['petal length (cm)'].min(), df['petal length (cm)'].max())]
```

```python
print(mean1)
print(mean2)
print(mean3)
```

```python
for j in range(100):
    k1=[]
    k2=[]
    k3=[]
    for i in range(df.shape[0]):
        x=np.sqrt(((df.loc[i]['sepal width (cm)']-mean1[0])**2)+((df.loc[i]['petal length (cm)']-mean1[1])**2))
        y=np.sqrt(((df.loc[i]['sepal width (cm)']-mean2[0])**2)+((df.loc[i]['petal length (cm)']-mean2[1])**2))
        z=np.sqrt(((df.loc[i]['sepal width (cm)']-mean3[0])**2)+((df.loc[i]['petal length (cm)']-mean3[1])**2))
        if min(x,y,z)==x:
            k1.append((df.loc[i]['sepal width (cm)'],df.loc[i]['petal length (cm)']))
        elif min(x,y,z)==y:
            k2.append((df.loc[i]['sepal width (cm)'],df.loc[i]['petal length (cm)']))
        elif min(x,y,z)==z:
            k3.append((df.loc[i]['sepal width (cm)'],df.loc[i]['petal length (cm)']))
    if k1:
        k1 = np.array(k1)
        mean1 = [np.mean(k1[:, 0]), np.mean(k1[:, 1])]

    if k2:
        k2 = np.array(k2)
        mean2 = [np.mean(k2[:, 0]), np.mean(k2[:, 1])]
    if k3:
        k3 = np.array(k3)
        mean3 = [np.mean(k3[:, 0]), np.mean(k3[:, 1])]

```

```python
print(mean1)
print(mean2)
print(mean3)
```

```python
print(len(k1))
print(len(k2))
print(len(k3))
```

```python
import matplotlib.pyplot as plt

x1, y1 = zip(*k1)
x2, y2 = zip(*k2)
x3, y3 = zip(*k3)

plt.scatter(x1, y1, color='blue', label='Cluster 1')
plt.scatter(x2, y2, color='green', label='Cluster 2')
plt.scatter(x3, y3, color='red', label='Cluster 3')

plt.scatter(mean1[0], mean1[1], color='black', marker='x', s=100 )
plt.scatter(mean2[0], mean2[1], color='black', marker='x', s=100)
plt.scatter(mean3[0], mean3[1], color='black', marker='x', s=100)

plt.legend()
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Two Clusters Scatter Plot")
plt.grid(True)

plt.show()
```
